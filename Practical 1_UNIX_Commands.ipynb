## 1. Basic UNIX Commands

Let's start by exploring our current environment, creating a directory structure, and managing simple text files.

```python
# Print working directory and list initial contents
!pwd
!ls -la
```

```python
# Create a directory, create a file inside, write content, and verify it
!mkdir -p practice
!echo "Hello UNIX! This is a clean demonstration." > practice/file.txt
!cat practice/file.txt
```

```python
# Inspect system information
!date
!whoami
!uname -a
```

## 2. Advanced UNIX Commands

Now, let's look at searching file contents, listing line counts, managing permissions, finding files, and creating compressed archives.

```python
# Search for a pattern and count lines in the target file
!grep "Hello" practice/file.txt
!wc -l practice/file.txt
```

```python
# Change file permissions and verify them
!chmod 644 practice/file.txt
!find practice -type f
```
